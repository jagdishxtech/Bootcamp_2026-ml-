```python
#Q19
import pandas as pd

data = {
    "Name": ["Aman", "Rahul", "Neha", "Riya", "Karan"],
    "Marks": [95, 82, 68, 45, 30]
}
df = pd.DataFrame(data)
def grade(mark):
    if mark >= 90:
        return "A"
    elif mark >= 75:
        return "B"
    elif mark >= 60:
        return "C"
    elif mark >= 40:
        return "D"
    else:
        return "F"
df["Grade"] = df["Marks"].apply(grade)
print(df)
```

```python
#Q20
import pandas as pd
data = {
    "Name": ["Aman", "Rahul", "Neha", "Riya"],
    "Marks": [450, 380, 420, 300]
}
df = pd.DataFrame(data)
def percentage(mark):
    return (mark / 500) * 100
df["Percentage"] = df["Marks"].apply(percentage)
print(df)
```
