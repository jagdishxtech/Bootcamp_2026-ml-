```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python

x=np.random.randint(2,20,5)
y=np.random.randint(2,20,5)
print(x,y,sep="\n")
plt.plot(x,y)
```

```python
df=pd.read_csv("/content/sharma-kohli.csv")
df=df.head(10)
x=df.iloc[:,1]
y=df.iloc[:,2]
plt.plot(x,y)
plt.xlabel("Rohit sharma")
plt.ylabel("Virat kohli")
```

```python
from matplotlib.lines import lineMarkers
plt.plot(df["index"],df['RG Sharma'],color="red",linestyle="solid",label="Rg",marker="o")
plt.plot(df["index"],df['V Kohli'],color="blue",linestyle="solid",label="vk",marker="o")
plt.title("score comparision between kohli and rohit sharma")
plt.xlabel("year")
plt.ylabel("score")
# if you want to see all the label of line then use the legend
plt.legend()
```

```python
# what if there is a outlier in data which will rediuce the varaition between the data
# in the case of outlier we use the ylim(min,max),xlim(min,max)
plt.plot(x,y)
plt.xlim(200,400)
plt.ylim(200,400)
plt.grid()
# there is main function which is used to show the graph
plt.show()
```
