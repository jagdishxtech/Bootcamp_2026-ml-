```python
marks = [45, 60, 72, 38, 90]

total = 0

for i in range(len(marks) - 1):
    total += marks[i]

average = total / len(marks)

if average > 50:
    print("Class passed")
else:
    print("Class failed")
print(average)
```

```python
def result(marks):
  avg=sum(marks)/len(marks)
  if avg>40:
    return "pass"
  else:
    return "fail"
marks = [45, 60, 72, 38, 90]
print(result(marks))
```

```python
df=pd.read_csv("/content/session4_dummy_students_data.csv.xls")
df.describe()
df["avg"] = df[["math", "science", "english"]].mean(axis=1)
df["result"] = "Fail"
df.loc[df["avg"] > 40, "result"] = "Pass"
print(df["avg"].mean())
print(df["avg"].min())
print(df["avg"].max())
print(df[df["avg"] == df["avg"].max()]["name"])
print(df[df["attendance_percent"]<75]["name"])
print(df[df["assignments_submitted"]<6]["name"])



```
